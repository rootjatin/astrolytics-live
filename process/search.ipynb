# 00 — Search and download astronomy data

Use this notebook to search MAST for a target, inspect matching observations, and download one or more FITS products.

This follows the same pattern as your Mars/JWST notebook: query observations, filter by instrument, rank products, and download locally. fileciteturn0file0


```python
from pathlib import Path
import numpy as np
from astropy.table import Table
from astro_utils import get_obs_table, pick_products_for_obs, download_one
```

```python
TARGET_NAME = "Mars"
INSTRUMENT = "NIRCAM"      # try "MIRI", "NIRSPEC", "WFC3", etc.
RADIUS = "0.3 deg"
PROGRAM_ID = 1415          # set to None to search more broadly
COLLECTION = "JWST"        # MAST obs_collection
DOWNLOAD_DIR = Path("data")

obs = get_obs_table(
    target=TARGET_NAME,
    instrument=INSTRUMENT,
    radius=RADIUS,
    program_id=PROGRAM_ID,
    collection=COLLECTION,
)
print(f"Found {len(obs)} matching observations")
obs[:10]
```

```python
# Show useful columns if present
wanted = [c for c in ["obsid", "target_name", "instrument_name", "t_exptime", "filters", "obs_collection"] if c in obs.colnames]
obs[wanted][:10]
```

```python
# Choose one observation row and inspect products
row_index = 0
row = obs[row_index]

prods = pick_products_for_obs(row)
print(f"Found {0 if prods is None else len(prods)} candidate products")
prods[:10] if prods is not None else None
```

```python
# Download a few files
max_to_download = 2
downloaded = []

if prods is not None:
    for pr in prods[:max_to_download]:
        path = download_one(pr, DOWNLOAD_DIR)
        downloaded.append(path)
        print("Downloaded:", path)

downloaded
```

### Adapt this notebook
- Change `TARGET_NAME` to your source
- Set `PROGRAM_ID=None` for a wider search
- Replace `INSTRUMENT` with the detector or instrument you need
- Increase `max_to_download` if you want multiple products
