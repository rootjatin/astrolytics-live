# 01 — Inspect a FITS file

Open a local FITS file, inspect its HDUs, print header values, and preview the main science data.


```
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
from astro_utils import read_fits_data, read_frame_cube, show_image
```

```
FITS_PATH = Path("data/your_file.fits")   # update this
HDU_NAME = "SCI"                          # or try 0 if needed
```

```
with fits.open(FITS_PATH) as hdul:
    hdul.info()
```

```
data, header = read_fits_data(FITS_PATH, ext=HDU_NAME)
print("Shape:", None if data is None else data.shape)
print("Datatype:", None if data is None else data.dtype)
```

```
# Show selected header keys
keys = ["OBJECT", "TELESCOP", "INSTRUME", "FILTER", "DATE-OBS", "EXPTIME", "BUNIT"]
{k: header.get(k) for k in keys}
```

```
# Handle 2D or cube data
if data.ndim == 2:
    show_image(data, title="2D science image")
else:
    cube, _ = read_frame_cube(FITS_PATH, ext=HDU_NAME)
    print("Cube shape:", cube.shape)
    show_image(np.nanmedian(cube, axis=0), title="Median image from cube")
    show_image(cube[0], title="First frame")
```

### What to look for
- Whether the science data is 2D or a time cube
- Whether `SCI`, `ERR`, `DQ`, or WCS-related extensions are present
- Header keys that matter for calibration and plotting
